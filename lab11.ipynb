```python
def edit_distance(source, destination):
    rows = len(destination) + 1
    cols = len(source) + 1

    dp = [[0 for _ in range(cols)] for _ in range(rows)]

    # Initialization
    for i in range(rows):
        dp[i][0] = i  # Vertical axis (Destination) - Deletions/Insertions
    for j in range(cols):
        dp[0][j] = j  # Horizontal axis (Source) - Deletions/Insertions

    # Fill DP table
    for i in range(1, rows):
        for j in range(1, cols):
            if destination[i - 1] == source[j - 1]:
                dp[i][j] = dp[i - 1][j - 1] # Cost 0 for match
            else:
                # New Costs:
                # Delete = 1, Insert = 1, Substitute = 2
                delete = dp[i][j - 1] + 1
                insert = dp[i - 1][j] + 1
                replace = dp[i - 1][j - 1] + 2

                dp[i][j] = min(insert, delete, replace)

    return dp

def print_table(dp, source, destination):
    print("\nDP Table (Source on Top, Destination on Left)\n")
    print("      ", end="")
    for ch in " " + source:
        print(f"{ch:4}", end="")
    print()

    for i in range(len(dp)):
        if i == 0:
            print("   ", end="")
        else:
            print(f"{destination[i-1]:3}", end="")
        for val in dp[i]:
            print(f"{val:4}", end="")
        print()

def get_operations(dp, source, destination):
    i = len(destination)
    j = len(source)
    operations = []

    while i > 0 or j > 0:
        if i > 0 and j > 0 and destination[i-1] == source[j-1] and dp[i][j] == dp[i-1][j-1]:
            operations.append(f"Ignore '{source[j-1]}'")
            i -= 1
            j -= 1
        elif i > 0 and j > 0 and dp[i][j] == dp[i-1][j-1] + 2:
            operations.append(f"Substitute '{source[j-1]}' → '{destination[i-1]}' (Cost 2)")
            i -= 1
            j -= 1
        elif j > 0 and dp[i][j] == dp[i][j-1] + 1:
            operations.append(f"Delete '{source[j-1]}' (Cost 1)")
            j -= 1
        else:
            operations.append(f"Insert '{destination[i-1]}' (Cost 1)")
            i -= 1

    operations.reverse()
    return operations

while True:
    print("\n--- New Experiment ---")
    source = input("Enter SOURCE word (or 'quit' to stop): ")
    if source.lower() == 'quit': break
    destination = input("Enter DESTINATION word: ")

    dp_table = edit_distance(source, destination)
    print_table(dp_table, source, destination)

    distance = dp_table[len(destination)][len(source)]
    print("\nMinimum Edit Distance =", distance)

    ops = get_operations(dp_table, source, destination)
    print("\nEdit Operations:")
    for step in ops:
        print("-", step)
```
